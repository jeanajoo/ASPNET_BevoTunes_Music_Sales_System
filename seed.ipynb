# Set Parameters

```python
import pyexcel as pe

# Declare constant PROJECT_NAME for C# using statements
PROJECT_NAME = "G6_BevoTunes"

seed_data = "Sp26 BevoTunes Seeding Data v2.xlsx"
```

## Genre

```python
header = ['GenreID', 'Name']
print(f"Length: {len(book['Genre']) - 1}")
lut_genre = {}
gs = ""
i=1
for row in book['Genre'].rows():
    if row == header:
        continue
    print("Processing Genre: ", row)
    lut_genre[row[1]] = row[0]
    gs += f"""
                Genre g{i} = new Genre() {{ GenreName = "{row[1]}" }};
                Genres.Add(g{i});
"""
    i += 1
   

print(gs)
```

## Artist

## Songs

```python
header = ['SongID', 'Name', 'Price', 'Album1', 'Album2', 'Album3', 'Artist', 'Genre1', 'Genre2', 'Genre3']
blank = ['', '', '', '', '', '', '', '', '', '']
print(f"Length: {len(book['Songs']) - 1}")
lut_song = {}
song_str = ""
i = 1
for row in book['Songs'].rows():
    if row == header or row == blank:
        continue
    print("Processing Songs: ", row)
    lut_song[row[1]] = row[0]
    song_str +=f"""
                Song s{i} = new Song()
                {{
                    SongTitle = "{row[1]}",
                    SongPrice = {row[2]}m
                }};
"""
    albums = [a for a in row[3:6] if a != '']
    for album in albums:
        song_str +=f"""                s{i}.Albums.Add(db.Albums.FirstOrDefault(a => a.AlbumName == "{album}"));
"""
    song_str +=f"""                s{i}.Artists.Add(db.Artists.FirstOrDefault(a => a.ArtistName == "{row[6]}"));
"""
    genres = [g for g in row[7:] if g != '']
    for genre in genres:
        song_str +=f"""                s{i}.Genres.Add(db.Genres.FirstOrDefault(g => g.GenreName == "{genre}"));
"""

    song_str +=f"""                Songs.Add(s{i});
"""
    i += 1
    
print(song_str)
    
   
```

## Promotions

```python
header = ['Artist', 'Song', 'Album', 'Promotion Type', 'Discount Quantity', 'Promotion Status']
blank = []
print(f"Length: {len(book['Promotion']) - 1}")
promotion_str = ""
i = 1
for row in book['Promotion'].rows():
    if row == header or row == blank:
        continue
    print("Processing Promotion: ", row)
    promotion_str +=f"""
                Promotion p{i} = new Promotion()
                {{
                    PromotionType = PromotionType.{row[3].upper()},
                    PromotionActive = {'true' if row[5] == "Active" else 'false'}
                }};
"""
    if row[0] != '':
        promotion_str +=f"""                p{i}.Artists.Add(db.Artists.FirstOrDefault(a => a.ArtistName == "{row[0]}"));
"""
    if row[1] != '':
        promotion_str +=f"""                p{i}.Songs.Add(db.Songs.FirstOrDefault(s => s.SongTitle == "{row[1]}"));
"""
    if row[2] != '':
        promotion_str +=f"""                p{i}.Albums.Add(db.Albums.FirstOrDefault(a => a.AlbumName == "{row[2]}"));
"""
    if row[3] == 'Discount':
        promotion_str +=f"""                p{i}.DiscountPrice = {row[4]}m;
"""
    promotion_str +=f"""                Promotions.Add(p{i});
"""

    i += 1
    
print(promotion_str)
```

## Orders

```python
header = ['Order_ID', 'Card_ID', 'Customer_Name', 'Date', 'Status', 'Song', 'Album', 'Price', 'GiftAddress']
blank = ['', '', '', '', '', '', '', '', '']
print(f"Length: {len(book['Orders']) - 1}")
lut_song = {}
order_str = ""
i = 0
curr_order_id = None
for row in book['Orders'].rows():
    if row == header or row == blank:
        continue
    print("Processing Orders: ", row)

    if curr_order_id != row[0] or curr_order_id == None:
        if curr_order_id != None:
            order_str += f"""
                Orders.Add(o{i});
"""
        i += 1
        curr_order_id  = row[0]

        fname = row[2].split(" ")[0]
        lname = row[2].split(" ")[1]

        if row[8] == '':
            rfname = fname
            rlname = lname
        else:
            rfname = row[8].split(" ")[0]
            rlname = row[8].split(" ")[1]

        order_str +=f"""
                Order o{i} = new Order()
                {{
                    Customer = customers.FirstOrDefault(u => u.FirstName == "{fname}" && u.LastName == "{lname}"),
                    Card = db.CustCreditCards.FirstOrDefault(c => c.CustCreditCardID == {row[1]}),
                    Recipient = customers.FirstOrDefault(u => u.FirstName == "{rfname}" && u.LastName == "{rlname}"),
                    OrderStatus = OrderStatus.{"InCart" if row[4] == "In_Cart" else row[4]}
                }};
"""
        if row[3] != '':
            order_str += f"""
                o{i}.OrderDate = DateTime.Parse("{row[3]}");
"""
    order_str += f"""
                o{i}.OrderDetails.Add(new OrderDetail()
                {{
                    PurchasedPrice = {row[7]}m,
"""
    if row[5] != '':
        order_str += f"""                    Song = db.Songs.FirstOrDefault(s => s.SongTitle == "{row[5]}"),
"""
    if row[6] != '':
        order_str += f"""                    Album = db.Albums.FirstOrDefault(a => a.AlbumName == "{row[6]}")
"""
    order_str += """                });
"""    
            
order_str += f"""
                Orders.Add(o{i});
"""
    
print(order_str)
    
   
```

## Reviews

```python
header = ['Reviewer', 'Item', 'Approver', 'Rating', 'ReviewText', 'DisputeStatus']
blank = ['', '', '', '', '', '']
print(f"Length: {len(book['Reviews']) - 1}")
review_str = ""
i = 1
for row in book['Reviews'].rows():
    if row == header or row == blank:
        continue
    print("Processing Reviews: ", row)

    rfname = row[0].split(" ")[0]
    rlname = row[0].split(" ")[1]

    review_str +=f"""
                Review r{i} = new Review()
                {{
                    Description = "{row[4]}",
                    Rating = {row[3]},
                    Status = ReviewStatus.{row[5]},
                    Reviewer = customers.FirstOrDefault(u => u.FirstName == "{rfname}" && u.LastName == "{rlname}"),
                }};
"""
    if row[2] != '':
        fname = row[2].split(" ")[0]
        lname = row[2].split(" ")[1]
        review_str +=f"""
                r{i}.Approver = adminsNemployees.FirstOrDefault(u => u.FirstName == "{fname}" && u.LastName == "{lname}");
"""
    review_str +=f"""
                Song songItem{i} = db.Songs.FirstOrDefault(s => s.SongTitle == "{row[1]}");
                if (songItem{i} != null)
                {{
                    r{i}.Song = songItem{i};
                }}

                Album albumItem{i} = db.Albums.FirstOrDefault(a => a.AlbumName == "{row[1]}");
                if (albumItem{i} != null)
                {{
                    r{i}.Album = albumItem{i};
                }}

                Artist artistItem{i} = db.Artists.FirstOrDefault(a => a.ArtistName == "{row[1]}");
                if (artistItem{i} != null)
                {{
                    r{i}.Artist = artistItem{i};
                }}

                Reviews.Add(r{i});
"""
    i += 1
    
print(review_str)
    
   
```
